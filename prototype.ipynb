# I. Modélisation 
## 1) Présentation du cadre
Considérons un graphe orienté $G = (S, A)$, où $S$ est un ensemble et $A \subset S^2$. Il y a une flèche du sommet $i \in S$ vers le sommet $j \in S$ si $(i, j) \in A$. Dans notre cas, les flèches représentent les routes et les sommets les points d'intersection. Pour tout couple $(i, j) \in S^2$, notons $P_{i,j}$ l'ensemble des chemins simples de $i$ vers $j$, c'est-à-dire les suites de sommets adjacents injectives débutant par $i$ et s'achevant par $j$. Pour simplifier les notations, supposons que $S = [|1, n|]$ où $n = |S|$.
Pour tout couple $(i, j) \in S^2$, notons $q_{i, j}$ la demande du sommet i vers le sommet j, autrement dit le nombre de voyageurs ayant pour origine $i$ et pour destination $j$ durant la période d'analyse.
La matrice $Q = (q_{i,j})_{1 \leq i, j \leq n}$ est appelée matrice d'origine-destination (OD).
Soit $a \in A$. Nous supposons que le temps de trajet de $a$, $t_a$, est uniquement fonction du flux sur $a$, $x_a$, c'est-à-dire le nombre de voyageurs passant par $a$. Pour tout couple $(i, j) \in S^2$ et tout chemin $k \in P_{i, j}$ les reliant, notons $f_k^{i,j}$ et $c_k^{i,j}$ les flux et temps de trajet, respectivement, du chemin $k$, en comptabilisant uniquement les voyageurs allant de $i$ à $j$ ; et pour tout $a \in A$, posons $\delta^{i,j}_{a,k} = 1$ si l'arête $a$ fait partie du chemin $k$ et $0$ sinon.
Nous avons alors les relations suivantes :
$$\forall (i, j) \in S^2, \forall k \in P_{i, j}, \, c_k^{i, j} = \sum_{a \in A} \delta^{i,j}_{a,k}t_a(x_a)$$
$$\forall a \in A, x_a = \sum_{(i, j) \in S^2}\sum_{a\in A}\delta^{i,j}_{a,k}f_k^{i,j}$$
$$\forall (i, j) \in S^2, q_{i, j} = \sum_{k \in P_{i, j}}f_{k}^{i, j}$$
La première exprime que le temps de trajet sur un certain chemin est la somme des temps de trajets sur les arêtes qu'il comprend, la deuxième que le flux sur une arête est la somme des flux sur tous les chemins qui passent à travers cet arc et la troisième que la demande d'un couple origine-demande est égale à la somme des flux sur les chemins les reliant.
Dans les différentes solutions du modèle que nous aborderons, les variables endogènes seront $f_k^{i,j}$ (et donc $x_a$), $Q$ est exogène.

## 2) Equilibres
L'optimum système est une configuration des flux minimisant la somme des temps de trajets des voyageurs. Il s'agit donc d'une solution au programme

\begin{align*}
&\min \sum_{a \in A} x_a t_a(x_a) \\
\text{s.c.} \quad & \\
& \forall (i, j) \in S^2, q_{i, j} = \sum_{k \in P_{i, j}}f_{k}^{i, j} \\
& \forall (i, j) \in S^2, \forall k \in P_{i, j}, f_{k}^{i, j} \geq 0 \\
\end{align*}

L'équilibre de Nash est moins immédiat à formuler. Informellement, dans cette situation, aucune déviation unilatérale n'est profitable pour l'agent qui la réalise (en considérant des agents infinitésimaux). Autrement dit, pour tout couple origine-destination $(i, j)$ et tout chemin $k \in P_{i, j}$ qui est emprunté (c'est-à-dire $f_{k}^{i, j} > 0$), le temps de trajet $c_k^{i,j}$ est une constante par rapport à $k$. En effet, si ce n'était pas le cas, des voyageurs auraient intérêt à quitter un chemin plus lent pour un plus rapide, et cela maintiendrait les autres contraintes. En d'autres termes, il existe des réels positifs $(u_{i, j})$ tels que
\begin{align*}
\forall (i, j) \in S^2, \forall k \in P_{i, j}, f_{k}^{i, j}(c_k^{i, j} - u_{i, j}) &= 0 \\
\forall (i, j) \in S^2, \forall k \in P_{i, j}, c_k^{i, j} - u_{i, j} &>= 0\\
\forall (i, j) \in S^2, \sum_{k \in P_{i, j}}f_{k}^{i, j} &= q_{i, j}  \\
\forall (i, j) \in S^2, \forall k \in P_{i, j}, f_{k}^{i, j} &\geq 0
\end{align*}
Il s'avère que ce sont exactement les conditions de KKT du programme de minimisation suivant : 

\begin{align*}
&\min \sum_{a \in A} \int_0^{x_a}t_a(y)dy\\
\text{s.c.} \quad & \\
& \forall (i, j) \in S^2, q_{i, j} = \sum_{k \in P_{i, j}}f_{k}^{i, j} \\
& \forall (i, j) \in S^2, \forall k \in P_{i, j}, f_{k}^{i, j} \geq 0 \\
\end{align*}

Alors que la première fonction objectif avait une interprétation simple, celle du second est un pur artéfact mathématique.
Sous les hypothèses de croissance de $t_a$ et de convexité de $x \to xt_a(x)$ pour tout $a \in A$ (qui seront vérifiées par les fonctions que nous emploierons), ces deux programmes sont des problèmes de minimasation convexes et admettent donc une unique solution.
